# 🧠 Understanding Memory in Educational AI

## This is the core of personalized tutoring with Mem0.

Let's see how Mem0 can automatically extract and store educationally relevant information from student interactions.

**Example Scenario**: A student shares their learning challenges and preferences.

*Notice how we're simply passing in a student's message, and Mem0 automatically extracts the relevant educational memories.*

```python
# 🎯 Example 1: Basic Educational Memory Storage

# Define a student ID and their learning message
student_id = "student-sam-1"

learning_message = "I'm really struggling with calculus derivatives, especially the chain rule. I understand basic algebra well, but calculus makes me anxious. I learn best with visual examples and step-by-step breakdowns."

messages = [
 {"role": "user", "content": learning_message}
]

# Store this educational context in memory
result = mem0_client.add(messages, user_id=student_id, version="v2", output_format="v1.1")

print("📚 Educational memories extracted:")
for memory in result['results']:
 print(f"• {memory['memory']}")
```

##  Step 2: Core Tutoring Functions with Memory Integration

Purpose: Build intelligent memory system that powers personalized tutoring

Key Functions:
- get_personalized_context(student_id, current_question) → Searches stored memories using mem0_client.search() with filters, returns formatted educational context
- store_educational_memory(messages, student_id) → Saves conversations using mem0_client.add() with custom_instructions for better memory filtering


```python
# 🧠 Step 2: Core tutoring functions with memory integration

from datetime import datetime

def get_personalized_context(student_id, current_question):
    """
    Retrieve relevant educational memories for personalized tutoring
    """
    try:
        # Search for memories relevant to the current question
        filters = {"OR": [{"user_id": student_id}]}
        memory_results = mem0_client.search(
            query=current_question,
            filters=filters,
            version="v2",
            output_format="v1.1"
        )

        # Build educational context
        context = f"Current time: {datetime.now().strftime('%Y-%m-%d at %H:%M')}\n\n"
        context += "🎓 PERSONALIZED TUTORING CONTEXT\n"
        context += "What I know about this student's learning:\n\n"

        if memory_results.get("results"):
            for i, result in enumerate(memory_results["results"][:4], 1):  # Top 4 memories
                memory_text = result["memory"]
                created_at = result.get("created_at", "")

                if created_at:
                    dt = datetime.fromisoformat(created_at.replace('Z', '+00:00'))
                    time_formatted = dt.strftime("%Y-%m-%d")
                    context += f"{i}. {memory_text} (noted on {time_formatted})\n"
                else:
                    context += f"{i}. {memory_text}\n"

        context += "\n🎯 TUTORING INSTRUCTIONS:\n"
        context += "- Provide personalized help based on the student's known strengths and weaknesses\n"
        context += "- Address any emotional concerns (anxiety, confidence) with empathy\n"
        context += "- Adapt your teaching style to their learning preferences\n"
        context += "- Reference previous topics when relevant\n"
        context += "- Encourage and build confidence\n\n"

        return context

    except Exception as e:
        return f"Context error: {str(e)}\n\n"


def store_educational_memory(messages, student_id, include_emotions=True):
    """
    Store educational interactions with custom instructions for better memory
    """
    try:
        if include_emotions:
            custom_instruction = (
                "Extract educational information including: learning topics, "
                "academic strengths/weaknesses, emotional states (anxiety, confidence), "
                "learning preferences, study habits, and specific subject difficulties."
            )
        else:
            custom_instruction = (
                "Focus on academic content: topics discussed, learning progress, "
                "and subject-specific strengths or challenges."
            )

        mem0_client.add(
            messages,
            user_id=student_id,
            custom_instructions=custom_instruction,
            version="v2",
            output_format="v1.1"
        )
        return True

    except Exception as e:
        print(f"Memory storage error: {str(e)}")
        return False


print("🧠 Core tutoring functions ready!")
```

##  Step 3: Main Tutoring Response Logic  

Purpose: Handle core conversation flow between student and AI tutor

Core Function:
- personalized_tutor_chat(message, student_id, chat_history) → Main orchestrator that calls get_personalized_context(), creates tutor_prompt, sends to gemini_client.models.generate_content(), stores interaction, and returns updated chat_history


```python
# 🤖 Step 3: Main tutoring response logic

def personalized_tutor_chat(message, student_id, chat_history):
    """
    Main function that handles personalized tutoring conversations
    """
    try:
        # Get personalized context from stored memories
        educational_context = get_personalized_context(student_id, message)

        # Create the tutoring prompt
        tutor_prompt = f"{educational_context}\n"
        tutor_prompt += f"Student Question: {message}\n\n"
        tutor_prompt += "As a personalized AI tutor, provide helpful, encouraging, and educational guidance:"

        # Get response from Gemini
        response = gemini_client.models.generate_content(
            model="gemini-2.5-flash",
            contents=tutor_prompt
        )

        tutor_response = response.text

        # Store this educational interaction in memory
        messages_to_store = [
            {"role": "user", "content": message},
            {"role": "assistant", "content": tutor_response}
        ]

        store_educational_memory(messages_to_store, student_id)

        # Update chat history
        chat_history.append((message, tutor_response))
        return chat_history, ""

    except Exception as e:
        error_msg = f"Tutor Error: {str(e)}"
        chat_history.append((message, error_msg))
        return chat_history, ""


print("🤖 Personalized tutoring logic ready!")
```

## Step 4: Memory Analysis and Student Profile Functions

Purpose: Create comprehensive learning profiles from stored educational data

Profile Function:
- get_student_learning_profile(student_id) → Fetches all memories with mem0_client.get_all(), categorizes into arrays like strengths[], weaknesses[], emotions[], uses keyword matching for classification, formats with timestamps


```python
# 📊 Step 4: Memory analysis and student profile functions


def get_student_learning_profile(student_id):
    """
    Generate a comprehensive learning profile from stored memories INCLUDING all stored memories
    """
    try:
        memories = mem0_client.get_all(user_id=student_id)

        if not memories:
            return "No learning history found. Start chatting to build your profile!"

        profile = "📚 PERSONALIZED LEARNING PROFILE\n"
        profile += "=" * 40 + "\n\n"

        # Categorize memories
        strengths = []
        weaknesses = []
        emotions = []
        preferences = []
        subjects = []

        for memory in memories:
            text = memory['memory'].lower()
            memory_text = memory['memory']

            # Simple categorization based on keywords
            if any(word in text for word in ['good at', 'excel', 'strong', 'understand']):
                strengths.append(memory_text)
            elif any(word in text for word in ['struggle', 'difficult', 'hard', 'weak']):
                weaknesses.append(memory_text)
            elif any(word in text for word in ['anxious', 'scared', 'confident', 'overwhelmed', 'stressed']):
                emotions.append(memory_text)
            elif any(word in text for word in ['learn best', 'prefer', 'visual', 'step-by-step']):
                preferences.append(memory_text)
            elif any(word in text for word in ['math', 'physics', 'chemistry', 'biology', 'calculus']):
                subjects.append(memory_text)

        # Build profile sections
        if strengths:
            profile += "💪 ACADEMIC STRENGTHS:\n"
            for strength in strengths[:3]:
                profile += f"  • {strength}\n"
            profile += "\n"

        if weaknesses:
            profile += "📈 AREAS FOR IMPROVEMENT:\n"
            for weakness in weaknesses[:3]:
                profile += f"  • {weakness}\n"
            profile += "\n"

        if emotions:
            profile += "🎭 EMOTIONAL INSIGHTS:\n"
            for emotion in emotions[:3]:
                profile += f"  • {emotion}\n"
            profile += "\n"

        if preferences:
            profile += "🎯 LEARNING PREFERENCES:\n"
            for pref in preferences[:3]:
                profile += f"  • {pref}\n"
            profile += "\n"

        # NEW SECTION: ALL STORED MEMORIES
        profile += "🧠 ALL STORED MEMORIES:\n"
        profile += "-" * 30 + "\n"

        for i, memory in enumerate(memories, 1):
            created_at = memory.get('created_at', '')
            memory_text = memory['memory']

            if created_at:
                # Format timestamp
                dt = datetime.fromisoformat(created_at.replace('Z', '+00:00'))
                time_formatted = dt.strftime("%m/%d %H:%M")
                profile += f"{i}. {memory_text}\n"
                profile += f"   📅 {time_formatted}\n\n"
            else:
                profile += f"{i}. {memory_text}\n\n"

        profile += "=" * 40 + "\n"
        profile += f"📊 Total memories: {len(memories)}\n"
        profile += f"📅 Profile updated: {datetime.now().strftime('%Y-%m-%d %H:%M')}"

        return profile

    except Exception as e:
        return f"Profile Error: {str(e)}"

print("📊 Enhanced student profile analysis with all memories ready!")
```

## Step 5: Create the Complete Gradio Interface

Purpose: Build user-friendly web interface with interactive components

Interface Components:
- create_personalized_tutor_interface() → Returns complete gr.Blocks() interface
- Creates chatbot = gr.Chatbot(), msg = gr.Textbox(), profile_display = gr.Textbox()
- Sets up event handlers: send_btn.click(), msg.submit(), refresh_profile_btn.click()
- All handlers defined inside with gr.Blocks() context


```python
# 🎨 Step 5: Create the complete Gradio interface

def create_personalized_tutor_interface():
    """
    Build the complete Gradio interface with event handlers inside the Blocks context
    """
    with gr.Blocks(title="Personalized AI Tutor", theme=gr.themes.Soft()) as demo:

        # Header and instructions
        gr.Markdown("""
        # 🎓 Personalized AI Tutor with Memory

        Welcome to your intelligent tutoring companion! This AI tutor remembers your learning journey,
        adapts to your needs, and provides personalized educational support.

        ## 🧪 Try These Example Interactions:

        1. "Hi! I'm struggling to understand how transformers work in deep learning and feel overwhelmed by the attention mechanism."

        2. "What are my learning strengths and weaknesses in machine learning topics?"

        3. "Can you explain LLMs using my preferred learning style? I'm feeling anxious about my upcoming AI exam."

        """)

        with gr.Row():
            # Main tutoring chat column
            with gr.Column(scale=2):
                gr.Markdown("### 💬 Chat with Your AI Tutor")
                chatbot = gr.Chatbot(
                    label="Tutoring Session",
                    height=500,
                    placeholder="Your personalized tutor will appear here..."
                )

                with gr.Row():
                    msg = gr.Textbox(
                        placeholder="Ask me anything about your studies...",
                        label="Your Question",
                        scale=4
                    )
                    student_id = gr.Textbox(
                        placeholder="student-001",
                        label="Student ID",
                        value="student-001",
                        scale=1
                    )

                with gr.Row():
                    send_btn = gr.Button("🎓 Ask Tutor", variant="primary")
                    clear_btn = gr.Button("🗑️ Clear Chat", variant="secondary")

            # Student profile and memory column
            with gr.Column(scale=1):
                gr.Markdown("### 📊 Your Learning Profile")
                profile_display = gr.Textbox(
                    label="Personalized Insights",
                    lines=15,
                    interactive=False,
                    placeholder="Your learning profile will appear here..."
                )

                with gr.Row():
                    refresh_profile_btn = gr.Button("🔄 Refresh Profile", variant="secondary")

                # Quick action buttons
                gr.Markdown("### ⚡ Quick Actions")

                with gr.Column():
                    strengths_btn = gr.Button("💪 My Strengths", size="sm")
                    weaknesses_btn = gr.Button("📈 Areas to Improve", size="sm")
                    anxiety_btn = gr.Button("😰 Feeling Anxious", size="sm")

        # Main chat functionality
        def send_message(message, student_id, history):
            if not message.strip():
                return history, ""
            return personalized_tutor_chat(message, student_id, history)

        def clear_chat():
            return []

        def refresh_student_profile(student_id):
            return get_student_learning_profile(student_id)

        # Quick action functions
        def ask_about_strengths(student_id, history):
            question = "What are my academic strengths based on our previous conversations?"
            return personalized_tutor_chat(question, student_id, history)

        def ask_about_weaknesses(student_id, history):
            question = "What are my learning weaknesses and how can I improve them?"
            return personalized_tutor_chat(question, student_id, history)

        def ask_about_yesterday(student_id, history):
            question = "What topics did we cover in our previous sessions?"
            return personalized_tutor_chat(question, student_id, history)

        def handle_anxiety(student_id, history):
            question = "I'm feeling anxious about my studies. Can you help me feel more confident?"
            return personalized_tutor_chat(question, student_id, history)

        # Connect main events
        send_btn.click(
            send_message,
            inputs=[msg, student_id, chatbot],
            outputs=[chatbot, msg]
        )

        msg.submit(
            send_message,
            inputs=[msg, student_id, chatbot],
            outputs=[chatbot, msg]
        )

        clear_btn.click(
            clear_chat,
            outputs=[chatbot]
        )

        refresh_profile_btn.click(
            refresh_student_profile,
            inputs=[student_id],
            outputs=[profile_display]
        )

        # Connect quick action buttons
        strengths_btn.click(
            ask_about_strengths,
            inputs=[student_id, chatbot],
            outputs=[chatbot, msg]
        )

        weaknesses_btn.click(
            ask_about_weaknesses,
            inputs=[student_id, chatbot],
            outputs=[chatbot, msg]
        )

        anxiety_btn.click(
            handle_anxiety,
            inputs=[student_id, chatbot],
            outputs=[chatbot, msg]
        )

        # Load student profile when interface loads
        demo.load(
            refresh_student_profile,
            inputs=[student_id],
            outputs=[profile_display]
        )

    return demo

print("🎨 Complete interface with event handlers created!")
```
